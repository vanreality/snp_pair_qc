## 20251112

20251112 8 PL samples, check its pair consistency with cfDNA samples

```python
import pandas as pd
import os
import glob
import re

data_dir = '/lustre1/cqyi/syfan/snp_nipt/data/220k_panel/qc_fastq'

fastq_dir_path = os.path.join(data_dir, '20251112/WES-251031')
pileup_dir_path = os.path.join(data_dir, '20251112/pileups')

meta = []

for sample in os.listdir(fastq_dir_path):
    fastq1 = os.path.join(fastq_dir_path, sample, f'{sample}_R1.fastq.gz')
    fastq2 = os.path.join(fastq_dir_path, sample, f'{sample}_R2.fastq.gz')
    pileup = os.path.join(pileup_dir_path, f'{sample.replace("V", "P").replace("M", "S")}_pileup.tsv.gz')

    if os.path.exists(fastq1) and os.path.exists(fastq2) and os.path.exists(pileup):
        meta.append({
            'sample': sample,
            'fastq1': fastq1,
            'fastq2': fastq2,
            'pileup': pileup
        })
        
meta_df = pd.DataFrame(meta)
meta_df.to_csv(f'{data_dir}/20251112_samplesheet.csv', index=False)
```

## 20251128

20251128 8 PL samples, check its pair consistency with cfDNA samples

Inhouse: inhouse seq with halved reagent (for wet lab condition exploration)

Out: out seq (for QC)

Out-downsampled: out seq downsampled (for comparison)

```python
import pandas as pd
import os
import glob
import re

dir_path = '/lustre1/cqyi/syfan/snp_nipt/data/220k_panel/raw_fastq/251128-PL-seq'
pileup_path = '/lustre1/cqyi/syfan/snp_nipt/results/220k_panel/summary/merge_pileup_hard_filter'

meta = []

for batch in ['inhouse', 'out', 'out_downsampled']:
    pattern = re.compile(r"(?P<sample>.+)_R1.fastq.gz$")

    for file in os.listdir(os.path.join(dir_path, f'fastq_{batch}')):
        match = pattern.match(file)

        if match:
            sample = match.group('sample')
            cfdna_sample = sample.replace('V', 'P')
            cfdna_pileup = os.path.join(pileup_path, f'{cfdna_sample}_pileup.tsv.gz')

            if os.path.exists(cfdna_pileup):
                meta.append({
                    'sample': sample,
                    'cfDNA_pileup': cfdna_pileup,
                    'fastq1': os.path.join(dir_path, f'fastq_{batch}', file),
                    'fastq2': os.path.join(dir_path, f'fastq_{batch}', file.replace('_R1.fastq.gz', '_R2.fastq.gz'))
                })
            
    meta_df = pd.DataFrame(meta)
    meta_df.to_csv(os.path.join(dir_path, f'{batch}_samplesheet.csv'), index=False)
    meta = []
```

```python
# Pileup samplesheet for rerun
import pandas as pd
import os
import glob
import re

dir_path = '/lustre1/cqyi/syfan/snp_nipt/results/pair_qc/20251128'
pileup_path = '/lustre1/cqyi/syfan/snp_nipt/results/220k_panel/summary/merge_pileup_hard_filter'

meta = []

for batch in ['inhouse', 'out', 'out_downsampled']:
    pattern = re.compile(r"(?P<sample>.+)_pileup.tsv.gz$")

    for file in os.listdir(os.path.join(dir_path, batch, 'snp_pileup')):
        match = pattern.match(file)

        if match:
            sample = match.group('sample')
            cfdna_sample = sample.replace('V', 'P')
            cfdna_pileup = os.path.join(pileup_path, f'{cfdna_sample}_pileup.tsv.gz')

            if os.path.exists(cfdna_pileup):
                meta.append({
                    'sample': sample,
                    'cfDNA_pileup': cfdna_pileup,
                    'pileup': os.path.join(dir_path, batch, 'snp_pileup', file)
                })
            
    meta_df = pd.DataFrame(meta)
    meta_df.to_csv(os.path.join(dir_path, f'{batch}_pileup_samplesheet.csv'), index=False)
    meta = []
```

```python
# Pileup samplesheet for heatmap
import pandas as pd
import os
import glob
import re

dir_path = '/lustre1/cqyi/syfan/snp_nipt/results/pair_qc/20251226'
pileup_path = '/lustre1/cqyi/syfan/snp_nipt/results/220k_panel/summary/merge_pileup_hard_filter'
samplesheet_path = '/lustre1/cqyi/syfan/snp_nipt/data/220k_panel/qc_fastq/20251226_samplesheet.csv'

meta = []

samplesheet = pd.read_csv(samplesheet_path)

for index, row in samplesheet.iterrows():
    sample = row['sample']
    # Use glob to find pileup files matching the sample
    pileup_files = glob.glob(f"{dir_path}/snp_pileup/{sample}*.tsv.gz")
    pileup = pileup_files[0] if pileup_files else None
    samplesheet.loc[index, 'pileup'] = pileup

samplesheet[['sample', 'cfDNA_pileup', 'pileup']].to_csv('/lustre1/cqyi/syfan/snp_nipt/results/pair_qc/20251226/20251226_samplesheet.csv', index=False)
```

## 20260203

IVF project 25 culture vs 26 tissue

```python
# Pileup samplesheet for heatmap
import pandas as pd
import os
import glob
import re

tissue_pileup_path = '/lustre1/cqyi/syfan/snp_nipt/results/pair_qc/20260203_ivf_qc/pileups/tissue'
culture_pileup_path = '/lustre1/cqyi/syfan/snp_nipt/results/pair_qc/20260203_ivf_qc/pileups/culture'
samplesheet_path = '/lustre1/cqyi/syfan/snp_nipt/results/pair_qc/20260203_ivf_qc/pileups/20260203_ivf_qc_samplesheet.csv'

pattern = re.compile(r"(?P<sample>.+)_pileup.tsv.gz$")
samplesheet = []

for pileup in os.listdir(tissue_pileup_path):
    match = pattern.match(pileup)
    if match:
        sample = match.group('sample')
        tissue_pileup = os.path.join(tissue_pileup_path, pileup)
        culture_pileup = os.path.join(culture_pileup_path, f'{sample.replace("B", "M")}_pileup.tsv.gz')
        if os.path.exists(tissue_pileup) and os.path.exists(culture_pileup):
            samplesheet.append({
                'sample': sample,
                'pileup': tissue_pileup,
                'cfDNA_pileup': culture_pileup
            })
        else:
            print(f"Missing pileup files for {sample}")

samplesheet.append({
    'sample': '4B',
    'pileup': os.path.join(tissue_pileup_path, '4B_pileup.tsv.gz'),
    'cfDNA_pileup': os.path.join(culture_pileup_path, '1M_pileup.tsv.gz')
})

samplesheet = pd.DataFrame(samplesheet)

samplesheet.to_csv(samplesheet_path, index=False)
```
